# The following 6 cells clean the data to account for format errors, missing values, missing value placeholders, etc. Run all six before proceeding.

```python
# @title Inspect, review, and clean precipitation + stream datasets
# Run AFTER loading DataFrames named `precip` and `stream`.
# Run AFTER the user has confirmed these are the correct .csvs

import numpy as np
from IPython.display import clear_output

# This grabs a snapshot of the files as they exist before we mess with them.
# It's terrible when the original data is lost to accidental overwrite.
raw_precip = precip.copy(deep=True)
raw_stream = stream.copy(deep=True)
RAW = {'Precipitation': raw_precip, 'Stream': raw_stream}

# This is a tricky part, and I can't think of a better method. We need to pre-define values we won't accept.
# I guess it's possible to ask it to compare each value to everthing else in that column and flag anything outside of the acceptable range...
# I don't like this second option because a value like -888.88 or -9999 is going to make "normal" a meaningless search, it risks artificial data flattening...
MISSING_TOKENS = {'', 'na', 'n/a', 'nan', 'null', 'none', 'missing', '.', '--',
                  '-888', '-888.0', '-888.00', '-888.88', '-888.888', '-999', '-9999'}
DATE_HINTS = ('date', 'year_month', 'timestamp', 'datetime')
ID_HINTS = ('id', 'identifier', 'code', 'site', 'station', 'watershed', 'sample')

```

```python
# @title Expected data types
# Set the standard for what the data *should* look like.
def normalize(series):
    s = series.astype('string').str.strip()
    return s.mask(s.str.lower().isin(MISSING_TOKENS).fillna(False))

def numeric_parse(series):
    return pd.to_numeric(series, errors='coerce')

#Setting up data verification so that is does not guess between 4/6/26 meaning Apr6 or Jun4, it sends it to the user
def date_parse(series, fmt):
    if fmt == 'Auto (unambiguous ISO only)':
        iso = series.str.fullmatch(r'\d{4}-\d{2}(-\d{2})?', na=False)
        safe = series.where(iso)
        return pd.to_datetime(safe, errors='coerce', format='mixed')
    return pd.to_datetime(series, format=fmt, errors='coerce')

# Does the data make sense with the column name?
def guess(col, original):
    label = str(col).lower()
    if any(k in label for k in DATE_HINTS):
        return 'Date'
    if label in ('year', 'month', 'day'):
        return 'Numeric'
    if any(k in label for k in ID_HINTS):
        return 'Text'
    s = normalize(original)
    good = s.dropna()
    if good.empty:
        return 'Text'
    nums = numeric_parse(good)
    if nums.notna().all():
        # Leading zeros is often an identifyer and shouldn't be removed without user approval
        if good.str.fullmatch(r'0\d+', na=False).any():
            return 'Text'
        return 'Numeric'
    if nums.notna().mean() >= .8:
        return 'Numeric'
    return 'Text'

```

```python
# @title Missing values default check
# Guide the user to be able to make appropriate changes to data type or leave them as they are.
# I like this approach because even though it's harder to make it work in the code, it keeps the user from
# making typos in the code trying to fix data types and it keeps the Thinking Machines from making assumptions.

# I see what Chat helped me build, but the styles and layouts and such... It gets heavy. It's giving dropdowns, though,
# with the suggested type as default and other data types as options.
suggestions = {}
settings = {}
for dataset, frame in RAW.items():
    for col in frame.columns:
        suggestions[(dataset, col)] = guess(col, frame[col])
        settings[(dataset, col)] = suggestions[(dataset, col)]

format_selector = widgets.Dropdown(
    options=['Auto (unambiguous ISO only)', '%Y-%m', '%Y-%m-%d', '%m/%d/%Y', '%d/%m/%Y'],
    value='Auto (unambiguous ISO only)', description='Date format:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='390px'))
extra_missing = widgets.Text(value='', description='Extra missing codes:',
    placeholder='Comma-separated, e.g. -777, ND', style={'description_width':'initial'},
    layout=widgets.Layout(width='490px'))
dataset_selector = widgets.Dropdown(options=list(RAW), description='Dataset:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='260px'))
search = widgets.Text(placeholder='Filter column names...', description='Search:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='340px'))
review_only = widgets.Checkbox(value=True, description='Review flagged only', indent=False)
column_selector = widgets.Dropdown(description='Column:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='390px'))
type_selector = widgets.Dropdown(options=['Numeric','Date','Text','Exclude'],
    description='Convert to:', style={'description_width':'initial'},
    layout=widgets.Layout(width='260px'))
column_info = widgets.HTML()
summary = widgets.HTML()

# Making the buttons to assign a function to further along.
preview_button = widgets.Button(description='Preview changes', icon='search', button_style='info')
apply_button = widgets.Button(description='Accept changes', icon='check', button_style='success')
results = widgets.Output()

# Now I have a data cleaning array of options, but they don't do anything yet.
```

```python
# @title Propose cleaned format
# This step "cleans" the data, but only as a proposed data set. This is what the machine thinks the data is.
def propose():
  # This is great. This was Chat's take on my idea. New missing value place holders can be entered.
  # If the data collector put "IDK" for unknowns, that can be entered here and turned to na.
  # Chat made this part of the dashboard while I had it as a separate step
    extra = {x.strip().lower() for x in extra_missing.value.split(',') if x.strip()}
    # Making bins for data that's good, needs review, or is bad
    cleaned, audit, issues = {}, [], []
    for dataset, frame in RAW.items():
        out = pd.DataFrame(index=frame.index)
        for col in frame.columns:
            kind = settings[(dataset, col)]
            if kind == 'Exclude':
                audit.append({'Dataset':dataset,'Column':col,'Action':'Excluded',
                              'Missing after':None,'Unparsed':0})
                continue
            source = frame[col]
            s = normalize(source)
            if extra:
                s = s.mask(s.str.lower().isin(extra).fillna(False))
            missing = int(s.isna().sum())
            bad = pd.Series(False, index=frame.index)
            if kind == 'Numeric':
                converted = numeric_parse(s).astype('float64')
                bad = s.notna() & converted.isna()
            elif kind == 'Date':
                converted = date_parse(s, format_selector.value)
                bad = s.notna() & converted.isna()
            else:
                converted = s
            if bad.any():
                examples = s.loc[bad].drop_duplicates().head(5).tolist()
                issues.append({'Dataset':dataset,'Column':col,'Type':kind,
                               'Unparsed':int(bad.sum()),'Examples':repr(examples)})
            out[col] = converted
            audit.append({'Dataset':dataset,'Column':col,'Action':kind,
                          'Missing after':missing,'Unparsed':int(bad.sum())})
        cleaned[dataset] = out
    return cleaned, pd.DataFrame(audit), pd.DataFrame(issues,
        columns=['Dataset','Column','Type','Unparsed','Examples'])

```

```python
# @title Make a summary
def flagged(dataset, col):
    s = normalize(RAW[dataset][col]); good = s.dropna()
    kind = settings[(dataset,col)]
    if kind == 'Numeric':
        return bool((good.notna() & numeric_parse(good).isna()).any())
    if kind == 'Date':
        return bool((good.notna() & date_parse(good,format_selector.value).isna()).any())
    return False

def update_summary():
    chunks = []
    for dataset, frame in RAW.items():
        nflag = sum(flagged(dataset,c) for c in frame.columns)
        chunks.append(f'<b>{dataset}</b>: {len(frame):,} rows, {len(frame.columns)} columns, '
                      f'{nflag} flagged columns')
    summary.value = ' &nbsp; | &nbsp; '.join(chunks)

def update_columns(*_):
    dataset = dataset_selector.value
    cols = [c for c in RAW[dataset].columns if search.value.lower() in str(c).lower()]
    if review_only.value:
        cols = [c for c in cols if flagged(dataset,c)]
    previous = column_selector.value
    column_selector.options = [(str(c),c) for c in cols]
    if previous in cols:
        column_selector.value = previous
    show_column()
    update_summary()

def show_column(*_):
    col = column_selector.value
    if col is None:
        column_info.value = '<i>No columns match. Uncheck “Review flagged only” to edit any column.</i>'
        type_selector.disabled = True
        return
    type_selector.disabled = False
    dataset = dataset_selector.value
    type_selector.unobserve(set_type, names='value')
    type_selector.value = settings[(dataset,col)]
    type_selector.observe(set_type, names='value')
    sample = normalize(RAW[dataset][col]).dropna().head(4).tolist()
    column_info.value = (f'<b>Suggested:</b> {suggestions[(dataset,col)]} &nbsp; '
                         f'<b>Current dtype:</b> {RAW[dataset][col].dtype} &nbsp; '
                         f'<b>Examples:</b> {str(sample)[:180]}')

def set_type(change):
    col = column_selector.value
    if col is not None:
        settings[(dataset_selector.value,col)] = change['new']
        # Don't rebuild the selector immediately; allow the user to finish editing.
        update_summary()

for control in (dataset_selector, search, review_only, format_selector):
    control.observe(update_columns, names='value')
column_selector.observe(show_column, names='value')

```

```python
# @title Manual Check
def preview(_):
    with results:
        clear_output(wait=True)
        proposed, audit, issues = propose()
        print('Preview only — working DataFrames have NOT changed.')
        print(f'Columns to convert/retain: {len(audit):,}')
        if len(issues):
            print(f'\nREVIEW REQUIRED: {len(issues)} columns contain values that cannot be parsed.')
            display(issues)
            print('Change the column type, correct the source, or add genuine missing-value tokens.')
        else:
            print('No unparseable values detected for selected conversions.')
        display(audit.groupby(['Dataset','Action'],dropna=False).size().rename('Columns').to_frame())

def apply(_):
    global precip, stream, cleaning_report
    with results:
        clear_output(wait=True)
        proposed, audit, issues = propose()
        if len(issues):
            print('NOT APPLIED. Resolve these unparseable values first:')
            display(issues)
            return
        precip = proposed['Precipitation']
        stream = proposed['Stream']
        cleaning_report = audit
        print('Cleaning accepted. precip and stream updated; raw_precip/raw_stream preserved.')
        print('Audit available as cleaning_report.')
        display(audit.groupby(['Dataset','Action'],dropna=False).size().rename('Columns').to_frame())

preview_button.on_click(preview)
apply_button.on_click(apply)
update_columns()

dashboard = widgets.VBox([
    widgets.HTML('<h3>Data cleaning review</h3>'), summary,
    widgets.HBox([dataset_selector,review_only]),
    widgets.HBox([search,column_selector]),
    column_info, type_selector,
    widgets.HTML('<b>Global cleaning options</b>'),
    format_selector, extra_missing,
    widgets.HBox([preview_button,apply_button]), results
])
display(dashboard)
```

# The precipitation and stream chemistry files are then combined into a single dataframe for analysis.

```python
# @title Shared Values
common_cols = precip.columns.intersection(stream.columns)

print("Shared columns:")
print(common_cols.tolist())
print("\nNumber of shared columns:", len(common_cols))

# I want the two files to match on year and month as the unique but shared values.
id_cols = ["Year", "Month"]

# Keep all columns except those I used to match. Those we don't need to double up on.
measurement_cols = [c for c in common_cols if c not in id_cols]

# Merge the files, but be sure to identify if the measure is from stream or precip.
paired = precip.merge(
    stream,
    on=id_cols,
    suffixes=("_precip", "_stream")
)

# This creates a dictionary to store keys (columns) and values (rows).
shared_counts = {}

# Then we count how many times they both have a non-*na* value.
for col in measurement_cols:
    shared_counts[col] = (
        paired[f"{col}_precip"].notna()
        & paired[f"{col}_stream"].notna()
    ).sum()

shared_counts

sum(shared_counts.values())

shared_counts_df = (
    pd.Series(shared_counts, name="paired_non_NA")
    .sort_values(ascending=False)
    .to_frame()
)

display(shared_counts_df)

print("Total paired non-NA cells:", shared_counts_df["paired_non_NA"].sum())
```

# Use this dropdown menu to select a chemical to compare between the precipitation and the stream.

```python
# @title Chemical Comparison { display-mode: "form" }

import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
import re

# Find chemical columns and their corresponding datasets
chemical_columns = {}

for col in legacy.columns:
    match = re.search(r"precipitation|precip|stream", str(col), re.I)

    if match:
        chemical = re.sub(
            r"precipitation|precip|stream", "", str(col),
            flags=re.I
        ).strip(" _-")

        source = "Precipitation" if "precip" in match.group().lower() else "Stream"

        chemical_columns.setdefault(chemical, {})[source] = col

# Only show chemicals found in BOTH datasets
chemicals = sorted(
    chem for chem, sources in chemical_columns.items()
    if "Precipitation" in sources and "Stream" in sources
)

chemical_dropdown = widgets.Dropdown(
    options=chemicals,
    description="Chemical:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

plot_output = widgets.Output()

def plot_chemical(change=None):
    with plot_output:
        plot_output.clear_output(wait=True)

        chemical = chemical_dropdown.value
        if chemical is None:
            print("No matching chemical columns found.")
            return

        columns = chemical_columns[chemical]

        precip_values = pd.to_numeric(
            legacy[columns["Precipitation"]], errors="coerce"
        )
        stream_values = pd.to_numeric(
            legacy[columns["Stream"]], errors="coerce"
        )

        # Use row number until a date column is identified
        x = range(len(legacy))

        plt.figure(figsize=(11, 5))

        plt.scatter(
            x, precip_values,
            label="Precipitation",
            color="royalblue",
            alpha=0.65,
            s=20
        )

        plt.scatter(
            x, stream_values,
            label="Stream",
            color="darkorange",
            alpha=0.65,
            s=20
        )

        plt.title(f"{chemical}: Precipitation vs Stream")
        plt.xlabel("Observation number")
        plt.ylabel("Chemical concentration")
        plt.legend()
        plt.grid(alpha=0.25)
        plt.tight_layout()
        plt.show()

chemical_dropdown.observe(plot_chemical, names="value")

display(chemical_dropdown, plot_output)
plot_chemical()
```

# Select if you want to view correlations between stream chemistry, precipitation chemistry, or both.

```python
# @title Chemical Correlation Heat Map { display-mode: "form" }

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display

# Select which dataset to examine
source_dropdown = widgets.Dropdown(
    options=["Both", "Precipitation", "Stream"],
    value="Both",
    description="Data source:",
    style={"description_width": "initial"}
)

heatmap_output = widgets.Output()

def show_heatmap(change=None):
    with heatmap_output:
        heatmap_output.clear_output(wait=True)

        # Select columns based on source
        source = source_dropdown.value

        if source == "Both":
            cols = [
                c for c in legacy.columns
                if "precip" in str(c).lower()
                or "stream" in str(c).lower()
            ]
        else:
            term = "precip" if source == "Precipitation" else "stream"
            cols = [
                c for c in legacy.columns
                if term in str(c).lower()
            ]

        # Convert selected columns to numeric
        data = legacy[cols].apply(
            pd.to_numeric, errors="coerce"
        )

        # Exclude columns with insufficient observations
        data = data.loc[:, data.count() >= 3]

        if data.shape[1] < 2:
            print("Not enough numeric columns for a correlation heat map.")
            return

        # Pearson correlation
        corr = data.corr()

        # Plot
        plt.figure(figsize=(
            max(8, len(corr) * 0.7),
            max(6, len(corr) * 0.6)
        ))

        sns.heatmap(
            corr,
            annot=len(corr) <= 15,
            cmap="coolwarm",
            center=0,
            vmin=-1,
            vmax=1,
            square=True,
            linewidths=0.4,
            cbar_kws={"label": "Pearson correlation (r)"}
        )

        plt.title(f"Chemical Correlations — {source}")
        plt.tight_layout()
        plt.show()

source_dropdown.observe(show_heatmap, names="value")

display(source_dropdown, heatmap_output)
show_heatmap()
```

```python

# @title 2. Define Binary Stream Outcome: Does the concentration or flux increase or decrease over the last measurement? { display-mode: "form" }

import pandas as pd
import numpy as np
import ipywidgets as widgets
from IPython.display import display

# Identify stream chemistry variables
stream_columns = [
    c for c in paired.columns
    if c.endswith("_stream")
    and c not in ["Year_Month_stream"]
]

# Include stream-only chemistry variables
stream_columns += [
    c for c in ["DIC_flux", "ANC_volwt"]
    if c in paired.columns
]

# Create readable dropdown labels
def chemical_label(col):
    return (
        col.replace("_flux_stream", "")
           .replace("_volwt_stream", "")
           .replace("_flux", "")
           .replace("_volwt", "")
           .replace("_stream", "")
    )

stream_dropdown = widgets.Dropdown(
    options=[(chemical_label(c), c) for c in stream_columns],
    description="Stream chemical:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

build_button = widgets.Button(
    description="Create Binary Outcome",
    button_style="success"
)

binary_output = widgets.Output()

def create_binary(button):
    global analysis_data, target_column

    with binary_output:
        binary_output.clear_output()

        target_column = stream_dropdown.value

        # Work on a copy, preserving paired
        df = paired.copy()

        # Build monthly date
        df["_date"] = pd.to_datetime(
            dict(
                year=df["Year"],
                month=df["Month"],
                day=1
            ),
            errors="coerce"
        )

        # Convert selected stream measurement to numeric
        df["_stream_value"] = pd.to_numeric(
            df[target_column], errors="coerce"
        )

        # Remove invalid dates and missing measurements
        df = df.dropna(
            subset=["_date", "_stream_value"]
        ).copy()

        # Sort chronologically
        df = df.sort_values("_date")

        # Calculate month-to-month change
        df["_previous"] = df["_stream_value"].shift(1)

        df["_change"] = (
            df["_stream_value"] - df["_previous"]
        )

        # Keep only consecutive calendar months
        previous_date = df["_date"].shift(1)

        consecutive = (
            (df["_date"].dt.year * 12 + df["_date"].dt.month)
            -
            (previous_date.dt.year * 12 + previous_date.dt.month)
        ) == 1

        df = df.loc[consecutive].copy()

        if df.empty:
            print("No consecutive months with valid data.")
            return

        # Create binary outcome
        df["stream_increase"] = (
            df["_change"] > 0
        ).astype(int)

        analysis_data = df

        print(f"Selected chemical: {chemical_label(target_column)}")
        print(f"Valid monthly comparisons: {len(df):,}")
        print()
        print("0 = No increase")
        print("1 = Increase")
        print()
        display(
            df["stream_increase"]
            .value_counts()
            .reindex([0, 1], fill_value=0)
            .rename_axis("Classification")
            .to_frame("Observations")
        )

        print("\nPreview:")
        display(
            df[[
                "_date",
                "_stream_value",
                "_previous",
                "_change",
                "stream_increase"
            ]].head()
        )

build_button.on_click(create_binary)

display(stream_dropdown, build_button, binary_output)
```

```python

# @title 4. Precipitation and Stream Relationships { display-mode: "form" }

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
from scipy.stats import spearmanr, mannwhitneyu
import statsmodels.api as sm

# Select precipitation chemistry or amount
precip_columns = [
    c for c in paired.columns
    if c.endswith("_precip") and c != "Year_Month_precip"
]

def precip_label(col):
    return col.replace("_flux_precip", "").replace(
        "_volwt_precip", ""
    ).replace("_precip", "")

precip_dropdown = widgets.Dropdown(
    options=[(precip_label(c), c) for c in precip_columns],
    description="Precipitation chemistry:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="450px")
)

lag_dropdown = widgets.Dropdown(
    options=[
        ("Same month", 0),
        ("1 month earlier", 1),
        ("2 months earlier", 2),
        ("3 months earlier", 3),
        ("6 months earlier", 6),
        ("12 months earlier", 12)
    ],
    description="Chemistry lag:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="450px")
)

# Independent explanatory-variable toggles
precip_amount_toggle = widgets.ToggleButton(
    value=False,
    description="Add precipitation amount",
    icon="plus",
    layout=widgets.Layout(width="250px")
)

flow_toggle = widgets.ToggleButton(
    value=False,
    description="Add streamflow",
    icon="plus",
    layout=widgets.Layout(width="250px")
)

def update_toggle(change, button, name):
    button.description = (
        f"Remove {name}" if change["new"] else f"Add {name}"
    )
    button.icon = "minus" if change["new"] else "plus"
    button.button_style = "success" if change["new"] else ""

precip_amount_toggle.observe(
    lambda change: update_toggle(
        change, precip_amount_toggle, "precipitation amount"
    ),
    names="value"
)

flow_toggle.observe(
    lambda change: update_toggle(
        change, flow_toggle, "streamflow"
    ),
    names="value"
)

analyze_button = widgets.Button(
    description="Analyze Relationship",
    button_style="primary"
)

relationship_output = widgets.Output()

def analyze_relationship(button):
    with relationship_output:
        relationship_output.clear_output(wait=True)

        if "analysis_data" not in globals():
            print("Run Cell 2 and create a binary outcome first.")
            return

        predictor = precip_dropdown.value
        lag = lag_dropdown.value

        # Create a calendar-indexed source table
        df = paired.copy()
        df["_date"] = pd.to_datetime(
            dict(year=df["Year"], month=df["Month"], day=1),
            errors="coerce"
        )
        df = df.dropna(subset=["_date"]).sort_values("_date")

        if df["_date"].duplicated().any():
            print("Duplicate months detected; resolve these first.")
            return

        # Lag chemistry and precipitation amount by calendar month
        history = df[["_date", predictor, "precip_mm"]].copy()
        history = history.rename(columns={
            predictor: "_precip_chem",
            "precip_mm": "_precip_amount"
        })
        history["_date"] += pd.DateOffset(months=lag)

        # Current-month stream outcome and flow
        current = analysis_data[
            ["_date", "_stream_value", "stream_increase", "flow_mm"]
        ].copy()

        data = current.merge(
            history, on="_date", how="left",
            validate="one_to_one"
        )

        variables = ["_precip_chem"]
        if precip_amount_toggle.value:
            variables.append("_precip_amount")
        if flow_toggle.value:
            variables.append("flow_mm")

        for c in variables + ["_stream_value"]:
            data[c] = pd.to_numeric(data[c], errors="coerce")

        data = data.dropna(
            subset=variables + ["_stream_value", "stream_increase"]
        ).copy()

        if len(data) < 8:
            print("Not enough complete observations for analysis.")
            return

        # Unadjusted inferential tests
        rho, p_corr = spearmanr(
            data["_precip_chem"], data["_stream_value"]
        )

        group0 = data.loc[
            data["stream_increase"] == 0, "_precip_chem"
        ]
        group1 = data.loc[
            data["stream_increase"] == 1, "_precip_chem"
        ]

        if len(group0) >= 3 and len(group1) >= 3:
            u_stat, p_mw = mannwhitneyu(
                group0, group1, alternative="two-sided"
            )
        else:
            u_stat, p_mw = np.nan, np.nan

        # Visualizations
        fig, axes = plt.subplots(1, 2, figsize=(13, 5))

        for cls, label, color in [
            (0, "Not increasing", "royalblue"),
            (1, "Increasing", "darkorange")
        ]:
            subset = data[data["stream_increase"] == cls]
            axes[0].scatter(
                subset["_precip_chem"],
                subset["_stream_value"],
                color=color, alpha=0.65, s=25,
                label=label
            )

        axes[0].set(
            xlabel=precip_label(predictor),
            ylabel=target_column,
            title="Precipitation vs Stream Measurement"
        )
        axes[0].legend()
        axes[0].grid(alpha=0.2)

        axes[1].boxplot(
            [group0, group1],
            tick_labels=["Not increasing", "Increasing"]
        )
        axes[1].set(
            title="Precipitation by Stream Outcome",
            ylabel=precip_label(predictor)
        )
        axes[1].grid(axis="y", alpha=0.2)

        plt.tight_layout()
        plt.show()

        print(f"Stream: {target_column}")
        print(f"Precipitation chemistry: {predictor}")
        print(f"Lag: {lag} month(s)")
        print(f"Complete observations: {len(data)}")
        print(f"Spearman rho: {rho:.3f}, p = {p_corr:.4g}")
        print(f"Mann-Whitney U: {u_stat:.2f}, p = {p_mw:.4g}")

        # Optional explanatory regression
        if precip_amount_toggle.value or flow_toggle.value:
            print("\nRegression accounting for selected variables:")

            X = sm.add_constant(
                data[variables], has_constant="add"
            )
            y = data["_stream_value"]

            if len(data) <= X.shape[1]:
                print("Too few observations for this model.")
                return

            try:
                model = sm.OLS(y, X).fit(
                    cov_type="HAC",
                    cov_kwds={
                        "maxlags": min(3, len(data) - 1)
                    }
                )

                results = pd.DataFrame({
                    "Coefficient": model.params,
                    "P-value": model.pvalues
                })

                display(results.round(4))
                print(f"R-squared: {model.rsquared:.3f}")
                print("HAC standard errors account for")
                print("some short-term serial correlation.")
            except Exception as exc:
                print(f"Regression could not be fitted: {exc}")
        else:
            print("\nNo additional explanatory variables selected.")

        print("\nThese analyses show associations, not causation.")

analyze_button.on_click(analyze_relationship)

display(
    precip_dropdown,
    lag_dropdown,
    widgets.HBox([
        precip_amount_toggle,
        flow_toggle
    ]),
    analyze_button,
    relationship_output
)
```

# Test a regression model for predictive power.

<center>$
P(Y=1)=\frac{1}{1+e^{-(\beta_0+\beta_1X_1+\cdots+\beta_kX_k)}}
$</center>

where Y = 1 means the chemical flux increased for the chosen X, the precipitation (mm/*t*) and flow (mm/*t*) variables.

```python

# @title 5. Logistic Regression Model { display-mode: "form" }

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    balanced_accuracy_score, roc_auc_score
)

# Use precipitation chemistry options from Cell 4
model_chem = widgets.Dropdown(
    options=[(precip_label(c), c) for c in precip_columns],
    description="Precipitation chemistry:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="450px")
)

model_lag = widgets.Dropdown(
    options=[
        ("Same month", 0),
        ("1 month earlier", 1),
        ("2 months earlier", 2),
        ("3 months earlier", 3),
        ("6 months earlier", 6),
        ("12 months earlier", 12)
    ],
    value=1,
    description="Chemistry lag:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="450px")
)

model_rain = widgets.Checkbox(
    value=False,
    description="Include precipitation amount"
)

model_flow = widgets.Checkbox(
    value=False,
    description="Include streamflow"
)

model_button = widgets.Button(
    description="Train Logistic Regression",
    button_style="success",
    icon="play"
)

model_output = widgets.Output()

def train_logistic_model(button):
    global logistic_model, logistic_results

    with model_output:
        model_output.clear_output(wait=True)

        if "analysis_data" not in globals():
            print("Run Cell 2 first.")
            return

        selected = model_chem.value
        lag = model_lag.value

        # Prepare full monthly precipitation history
        history = paired.copy()
        history["_date"] = pd.to_datetime(
            dict(year=history["Year"],
                 month=history["Month"], day=1),
            errors="coerce"
        )
        history = history.dropna(subset=["_date"])

        if history["_date"].duplicated().any():
            print("Duplicate dates found; resolve before modeling.")
            return

        predictors = [selected]
        if model_rain.value:
            predictors.append("precip_mm")
        if model_flow.value:
            predictors.append("flow_mm")

        # Lag precipitation variables by calendar month
        precip_predictors = [
            c for c in predictors if c != "flow_mm"
        ]

        lagged = history[["_date"] + precip_predictors].copy()
        lagged["_date"] = (
            lagged["_date"] + pd.DateOffset(months=lag)
        )

        # Use valid monthly stream outcomes
        outcome_cols = ["_date", "stream_increase"]
        if model_flow.value:
            outcome_cols.append("flow_mm")

        data = analysis_data[outcome_cols].merge(
            lagged, on="_date", how="left",
            validate="one_to_one"
        )

        data = data.sort_values("_date").copy()

        # Numeric conversion
        for col in predictors:
            data[col] = pd.to_numeric(
                data[col], errors="coerce"
            )

        # Require observed chemistry predictor and outcome
        data = data.dropna(
            subset=[selected, "stream_increase"]
        ).copy()

        if len(data) < 30:
            print("At least 30 usable observations are required.")
            return

        X = data[predictors]
        y = data["stream_increase"].astype(int)

        # Chronological 80/20 split
        split = int(len(data) * 0.8)

        X_train, X_test = X.iloc[:split], X.iloc[split:]
        y_train, y_test = y.iloc[:split], y.iloc[split:]

        if y_train.nunique() < 2:
            print("Training data contain only one outcome class.")
            return

        if y_test.nunique() < 2:
            print("Test data contain only one outcome class.")
            print("Choose another chemical or inspect the time series.")
            return

        # Imputation and scaling learned only from training data
        logistic_model = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("classifier", LogisticRegression(
                max_iter=2000,
                class_weight="balanced",
                random_state=42
            ))
        ])

        logistic_model.fit(X_train, y_train)

        probabilities = logistic_model.predict_proba(X_test)[:, 1]
        predictions = (probabilities >= 0.5).astype(int)

        balanced_acc = balanced_accuracy_score(
            y_test, predictions
        )
        auc = roc_auc_score(y_test, probabilities)

        # Coefficients refer to standardized predictors
        coefficients = logistic_model.named_steps[
            "classifier"
        ].coef_[0]

        coefficient_table = pd.DataFrame({
            "Predictor": predictors,
            "Coefficient": coefficients,
            "Odds ratio": np.exp(coefficients)
        })

        logistic_results = data.iloc[split:].copy()
        logistic_results["Predicted probability"] = probabilities
        logistic_results["Predicted class"] = predictions

        print(f"Stream outcome: {target_column}")
        print(f"Precipitation chemistry: {selected}")
        print(f"Precipitation lag: {lag} month(s)")
        print(f"Training observations: {len(X_train)}")
        print(f"Testing observations: {len(X_test)}")
        print()
        print(f"Balanced accuracy: {balanced_acc:.3f}")
        print(f"ROC AUC: {auc:.3f}")

        print("\nModel coefficients:")
        display(coefficient_table.round(3))

        fig, axes = plt.subplots(1, 2, figsize=(13, 5))

        # Confusion matrix
        cm = confusion_matrix(
            y_test, predictions, labels=[0, 1]
        )
        ConfusionMatrixDisplay(
            cm,
            display_labels=["No increase", "Increase"]
        ).plot(ax=axes[0], colorbar=False)

        axes[0].set_title("Test Set Predictions")

        # Predicted probabilities over time
        axes[1].plot(
            logistic_results["_date"],
            probabilities,
            color="steelblue",
            linewidth=1.5,
            label="Predicted probability"
        )
        axes[1].scatter(
            logistic_results["_date"],
            y_test,
            color="darkorange",
            s=20,
            alpha=0.6,
            label="Observed outcome"
        )
        axes[1].axhline(
            0.5, color="gray", linestyle="--",
            label="Classification threshold"
        )
        axes[1].set(
            title="Probability of Stream Increase",
            xlabel="Date",
            ylabel="Probability / observed class",
            ylim=(-0.05, 1.05)
        )
        axes[1].legend()
        axes[1].grid(alpha=0.2)

        plt.tight_layout()
        plt.show()

model_button.on_click(train_logistic_model)

display(
    model_chem,
    model_lag,
    model_rain,
    model_flow,
    model_button,
    model_output
)
```

```python

# @title 6. Model Evaluation and Interpretation { display-mode: "form" }

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, balanced_accuracy_score, roc_auc_score,
    confusion_matrix, ConfusionMatrixDisplay
)

PRIMARY_METRIC = "Balanced accuracy"

evaluate_button = widgets.Button(
    description="Compare Classification Models",
    button_style="success",
    icon="chart-bar"
)

evaluation_output = widgets.Output()

def evaluate_models(button):
    global model_comparison, best_model, best_model_name

    with evaluation_output:
        evaluation_output.clear_output(wait=True)

        if "analysis_data" not in globals():
            print("Run Cell 2 first.")
            return

        # Read the researcher's choices from Cell 5
        selected = model_chem.value
        lag = model_lag.value

        predictors = [selected]

        if model_rain.value:
            predictors.append("precip_mm")

        if model_flow.value:
            predictors.append("flow_mm")

        # Build monthly precipitation history
        history = paired.copy()
        history["_date"] = pd.to_datetime(
            dict(
                year=history["Year"],
                month=history["Month"],
                day=1
            ),
            errors="coerce"
        )

        history = history.dropna(subset=["_date"])
        history = history.sort_values("_date")

        if history["_date"].duplicated().any():
            print("Duplicate dates found. Resolve these first.")
            return

        precip_predictors = [
            c for c in predictors if c != "flow_mm"
        ]

        lagged = history[
            ["_date"] + precip_predictors
        ].copy()

        lagged["_date"] += pd.DateOffset(months=lag)

        outcome_cols = ["_date", "stream_increase"]

        if model_flow.value:
            outcome_cols.append("flow_mm")

        data = analysis_data[outcome_cols].merge(
            lagged,
            on="_date",
            how="left",
            validate="one_to_one"
        )

        data = data.sort_values("_date").copy()

        for col in predictors:
            data[col] = pd.to_numeric(
                data[col], errors="coerce"
            )

        # Match Cell 5's usable observations
        data = data.dropna(
            subset=[selected, "stream_increase"]
        ).copy()

        if len(data) < 30:
            print("At least 30 observations are required.")
            return

        X = data[predictors]
        y = data["stream_increase"].astype(int)

        split = int(len(data) * 0.8)

        X_train = X.iloc[:split]
        X_test = X.iloc[split:]

        y_train = y.iloc[:split]
        y_test = y.iloc[split:]

        if y_train.nunique() < 2 or y_test.nunique() < 2:
            print("Both classes must occur in training and test data.")
            return

        if X_train.isna().all().any():
            print("A predictor has no training observations:")
            print(X_train.columns[X_train.isna().all()].tolist())
            return

        # Same split, same features, same preprocessing
        models = {
            "Baseline": DummyClassifier(
                strategy="most_frequent"
            ),
            "Logistic Regression": LogisticRegression(
                max_iter=2000,
                class_weight="balanced",
                random_state=42
            ),
            "Decision Tree": DecisionTreeClassifier(
                max_depth=4,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=42
            ),
            "Random Forest": RandomForestClassifier(
                n_estimators=200,
                max_depth=5,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=42
            )
        }

        results = []
        fitted_models = {}
        predictions_by_model = {}

        for name, classifier in models.items():

            steps = [
                ("imputer", SimpleImputer(strategy="median"))
            ]

            if name == "Logistic Regression":
                steps.append(("scaler", StandardScaler()))

            steps.append(("classifier", classifier))

            pipeline = Pipeline(steps)
            pipeline.fit(X_train, y_train)

            pred = pipeline.predict(X_test)
            prob = pipeline.predict_proba(X_test)[:, 1]

            results.append({
                "Model": name,
                "Accuracy": accuracy_score(y_test, pred),
                "Precision": precision_score(
                    y_test, pred, zero_division=0
                ),
                "Recall": recall_score(
                    y_test, pred, zero_division=0
                ),
                "F1": f1_score(
                    y_test, pred, zero_division=0
                ),
                "Balanced accuracy": balanced_accuracy_score(
                    y_test, pred
                ),
                "ROC AUC": roc_auc_score(y_test, prob)
            })

            fitted_models[name] = pipeline
            predictions_by_model[name] = pred

        model_comparison = pd.DataFrame(results)

        model_comparison = model_comparison.sort_values(
            PRIMARY_METRIC, ascending=False
        ).reset_index(drop=True)

        # Exclude baseline from model selection
        candidates = model_comparison[
            model_comparison["Model"] != "Baseline"
        ]

        best_model_name = candidates.iloc[0]["Model"]
        best_model = fitted_models[best_model_name]

        print("MODEL PERFORMANCE")
        print(f"Stream measurement: {target_column}")
        print(f"Predictors: {', '.join(predictors)}")
        print(f"Precipitation lag: {lag} month(s)")
        print(f"Training months: {len(X_train)}")
        print(f"Testing months: {len(X_test)}")
        print(f"Primary metric: {PRIMARY_METRIC}")
        print()

        display(model_comparison.round(3))

        # Compare performance visually
        fig, axes = plt.subplots(1, 2, figsize=(13, 5))

        ordered = model_comparison.iloc[::-1]

        axes[0].barh(
            ordered["Model"],
            ordered[PRIMARY_METRIC]
        )

        axes[0].set_xlim(0, 1)
        axes[0].set_xlabel(PRIMARY_METRIC)
        axes[0].set_title("Model Performance")
        axes[0].axvline(
            0.5, color="gray", linestyle="--"
        )

        # Confusion matrix for selected model
        cm = confusion_matrix(
            y_test,
            predictions_by_model[best_model_name],
            labels=[0, 1]
        )

        ConfusionMatrixDisplay(
            cm,
            display_labels=["No increase", "Increase"]
        ).plot(ax=axes[1], colorbar=False)

        axes[1].set_title(
            f"Confusion Matrix: {best_model_name}"
        )

        plt.tight_layout()
        plt.show()

        # Automatic interpretation
        best_score = candidates.iloc[0][PRIMARY_METRIC]

        baseline_score = model_comparison.loc[
            model_comparison["Model"] == "Baseline",
            PRIMARY_METRIC
        ].iloc[0]

        print("\nMODEL INTERPRETATION")
        print(f"Selected model: {best_model_name}")
        print(f"Balanced accuracy: {best_score:.3f}")
        print(f"Baseline balanced accuracy: {baseline_score:.3f}")

        if best_score > baseline_score:
            print("The selected model outperformed the baseline.")
        else:
            print("The selected model did not outperform the baseline.")

        print("\nSelection rule:")
        print("Highest test-set balanced accuracy among")
        print("the three classification models.")

        print("\nImportant:")
        print("These are observational associations.")
        print("Test-set selection is exploratory; confirm")
        print("performance with rolling time-series validation.")

evaluate_button.on_click(evaluate_models)

display(evaluate_button, evaluation_output)
```
