# Hubbard Brook Precipitation–Stream Chemistry Analysis
This is an interactive Python based workflow for exploring relationships between precipitation chemistry, water movement, and changes in stream chemistry using NFS Hubbard Brook Experimental Forest observations. The ideas for the research questions, the functionality of the notebook, the data analysis tools, and the typos are the author's. The coode was written with the assistance of ChatGPT from OpenAI.

###Purpose
This notebook uses monthly precipitation and stream measurements to visualize chemical changes, examine cimmediate and lagged associations, and evaluate supervised classification models that predict whether a selected stream measurement increases relative to the preceding calendar month.
This user friendly notebook is designed for researchers and uses dropdowns, checkboxes, and buttons in lieu of editing Python code. It is an exploratory research tool, not a causal inference or operational forecasting system. The user friendly aspect is intended to allow a researcher to casually browse associations prior to more rigorous efforts.

###Data and structure
The combined, cleaned pandas DataFrame is named *paired.* It includes Year, Month, precipitation and stream month identifiers, precipitation amount (precip_mm), streamflow (flow_mm), precipitation chemistry variables (such as Na_flux_precip), and stream chemistry variables (such as NH4_flux_stream). Stream-only variables include DIC_flux (dissolved inorganic carbon) and ANC_volwt (acid neutralizing capacity), as these don't really measure well, nor are they stable, in precipitation.

Missing-value placeholders such as -888 are converted to missing values during preprocessing. The current paired dataset has no explicit watershed identifier; the analysis therefore assumes a single monthly time series. Confirm this assumption against the source files before interpreting results. To include all watersheds in a single analysis, a watershed identifier would need to be added to each .csv downloaded from HBEF.

Measurement terminology: Many columns represent flux, not concentration. Flux is the amount of a chemical that passes a given point over a given time as opposed to concentration: an amount in a given volume. Variables with _volwt are "volume-weighted measurements", concentrations, and should not be described as flux.

###Outcome definition
This notebook looks at whether a chemical metric increased or not. While not increasing is most often occuring with decreasing, they are not synonymous.

For a stream measurement $S_t$, define the monthly change as:
$
\Delta S_t = S_t - S_{t-1}
$
The binary target is:
<center>$
Y_t = \begin{cases}
1 & \text{if } \Delta S_t > 0 \\
0 & \text{if } \Delta S_t \leq 0
\end{cases}
$</center>


Missing months are not treated as adjacent observations.  
Statistical and machine-learning methods
- Spearman rank correlation: explores monotonic association between selected precipitation and stream measurements.
- Mann–Whitney U test: compares precipitation measurements in months classified as increasing versus not increasing.
- Multiple linear regression: estimates associations with the continuous stream measurement while including selected water volume covariates
- Logistic regression: estimates the probability of a binary increase
- Decision tree and random forest: provide alternative models

###Lags and explanatory variables
Delayed responses to precipitation inputs have been explored in this author's own research and with the multi-state LoVoTECS stream observation network.

The interface supports precipitation lags of 0, 1, 2, 3, 6, or 12 calendar months.

Model evaluation
The current workflow splits observations chronologically: the earliest 80% for training and the latest 20% for testing.

Reported metrics are accuracy, precision, recall, F1, balanced accuracy, and ROC AUC. Balanced accuracy is the predefined primary metric for selecting the best classifier because the outcome classes may be unevenly represented.

###Limitations and cautions
- Observational associations do not establish causation.
- Monthly observations are serially correlated and may exhibit seasonality and long-term trends.
- Stream chemical flux can depend strongly on flow; interpret chemistry and hydrology jointly.
- The single 80/20 chronological split may be sensitive to the chosen time period.
- Missing-data patterns and original watershed coverage need verification.
- Trying many chemical/lag combinations increases the chance of apparently strong results arising by chance.
