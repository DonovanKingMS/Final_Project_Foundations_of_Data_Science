```python
# @title Inspect, review, and clean precipitation + stream datasets
# Run AFTER loading DataFrames named `precip` and `stream`.
# Run AFTER the user has confirmed these are the correct .csvs

import numpy as np
from IPython.display import clear_output

# This grabs a snapshot of the files as they exist before we mess with them.
# It's terrible when the original data is lost to accidental overwrite.
raw_precip = precip.copy(deep=True)
raw_stream = stream.copy(deep=True)
RAW = {'Precipitation': raw_precip, 'Stream': raw_stream}

# This is a tricky part, and I can't think of a better method. We need to pre-define values we won't accept.
# I guess it's possible to ask it to compare each value to everthing else in that column and flag anything outside of the acceptable range...
# I don't like this second option because a value like -888.88 or -9999 is going to make "normal" a meaningless search, it risks artificial data flattening...
MISSING_TOKENS = {'', 'na', 'n/a', 'nan', 'null', 'none', 'missing', '.', '--',
                  '-888', '-888.0', '-888.00', '-888.88', '-888.888', '-999', '-9999'}
DATE_HINTS = ('date', 'year_month', 'timestamp', 'datetime')
ID_HINTS = ('id', 'identifier', 'code', 'site', 'station', 'watershed', 'sample')

```

```python
# Set the standard for what the data *should* look like.
def normalize(series):
    s = series.astype('string').str.strip()
    return s.mask(s.str.lower().isin(MISSING_TOKENS).fillna(False))

def numeric_parse(series):
    return pd.to_numeric(series, errors='coerce')

#Setting up data verification so that is does not guess between 4/6/26 meaning Apr6 or Jun4, it sends it to the user
def date_parse(series, fmt):
    if fmt == 'Auto (unambiguous ISO only)':
        iso = series.str.fullmatch(r'\d{4}-\d{2}(-\d{2})?', na=False)
        safe = series.where(iso)
        return pd.to_datetime(safe, errors='coerce', format='mixed')
    return pd.to_datetime(series, format=fmt, errors='coerce')

# Does the data make sense with the column name?
def guess(col, original):
    label = str(col).lower()
    if any(k in label for k in DATE_HINTS):
        return 'Date'
    if label in ('year', 'month', 'day'):
        return 'Numeric'
    if any(k in label for k in ID_HINTS):
        return 'Text'
    s = normalize(original)
    good = s.dropna()
    if good.empty:
        return 'Text'
    nums = numeric_parse(good)
    if nums.notna().all():
        # Leading zeros is often an identifyer and shouldn't be removed without user approval
        if good.str.fullmatch(r'0\d+', na=False).any():
            return 'Text'
        return 'Numeric'
    if nums.notna().mean() >= .8:
        return 'Numeric'
    return 'Text'

```

```python
# Guide the user to be able to make appropriate changes to data type or leave them as they are.
# I like this approach because even though it's harder to make it work in the code, it keeps the user from
# making typos in the code trying to fix data types and it keeps the Thinking Machines from making assumptions.

# I see what Chat helped me build, but the styles and layouts and such... It gets heavy. It's giving dropdowns, though,
# with the suggested type as default and other data types as options.
suggestions = {}
settings = {}
for dataset, frame in RAW.items():
    for col in frame.columns:
        suggestions[(dataset, col)] = guess(col, frame[col])
        settings[(dataset, col)] = suggestions[(dataset, col)]

format_selector = widgets.Dropdown(
    options=['Auto (unambiguous ISO only)', '%Y-%m', '%Y-%m-%d', '%m/%d/%Y', '%d/%m/%Y'],
    value='Auto (unambiguous ISO only)', description='Date format:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='390px'))
extra_missing = widgets.Text(value='', description='Extra missing codes:',
    placeholder='Comma-separated, e.g. -777, ND', style={'description_width':'initial'},
    layout=widgets.Layout(width='490px'))
dataset_selector = widgets.Dropdown(options=list(RAW), description='Dataset:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='260px'))
search = widgets.Text(placeholder='Filter column names...', description='Search:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='340px'))
review_only = widgets.Checkbox(value=True, description='Review flagged only', indent=False)
column_selector = widgets.Dropdown(description='Column:',
    style={'description_width':'initial'}, layout=widgets.Layout(width='390px'))
type_selector = widgets.Dropdown(options=['Numeric','Date','Text','Exclude'],
    description='Convert to:', style={'description_width':'initial'},
    layout=widgets.Layout(width='260px'))
column_info = widgets.HTML()
summary = widgets.HTML()

# Making the buttons to assign a function to further along.
preview_button = widgets.Button(description='Preview changes', icon='search', button_style='info')
apply_button = widgets.Button(description='Accept changes', icon='check', button_style='success')
results = widgets.Output()

# Now I have a data cleaning array of options, but they don't do anything yet.
```

```python
# This step "cleans" the data, but only as a proposed data set. This is what the machine thinks the data is.
def propose():
  # This is great. This was Chat's take on my idea. New missing value place holders can be entered.
  # If the data collector put "IDK" for unknowns, that can be entered here and turned to na.
  # Chat made this part of the dashboard while I had it as a separate step
    extra = {x.strip().lower() for x in extra_missing.value.split(',') if x.strip()}
    # Making bins for data that's good, needs review, or is bad
    cleaned, audit, issues = {}, [], []
    for dataset, frame in RAW.items():
        out = pd.DataFrame(index=frame.index)
        for col in frame.columns:
            kind = settings[(dataset, col)]
            if kind == 'Exclude':
                audit.append({'Dataset':dataset,'Column':col,'Action':'Excluded',
                              'Missing after':None,'Unparsed':0})
                continue
            source = frame[col]
            s = normalize(source)
            if extra:
                s = s.mask(s.str.lower().isin(extra).fillna(False))
            missing = int(s.isna().sum())
            bad = pd.Series(False, index=frame.index)
            if kind == 'Numeric':
                converted = numeric_parse(s).astype('float64')
                bad = s.notna() & converted.isna()
            elif kind == 'Date':
                converted = date_parse(s, format_selector.value)
                bad = s.notna() & converted.isna()
            else:
                converted = s
            if bad.any():
                examples = s.loc[bad].drop_duplicates().head(5).tolist()
                issues.append({'Dataset':dataset,'Column':col,'Type':kind,
                               'Unparsed':int(bad.sum()),'Examples':repr(examples)})
            out[col] = converted
            audit.append({'Dataset':dataset,'Column':col,'Action':kind,
                          'Missing after':missing,'Unparsed':int(bad.sum())})
        cleaned[dataset] = out
    return cleaned, pd.DataFrame(audit), pd.DataFrame(issues,
        columns=['Dataset','Column','Type','Unparsed','Examples'])

```

```python
def flagged(dataset, col):
    s = normalize(RAW[dataset][col]); good = s.dropna()
    kind = settings[(dataset,col)]
    if kind == 'Numeric':
        return bool((good.notna() & numeric_parse(good).isna()).any())
    if kind == 'Date':
        return bool((good.notna() & date_parse(good,format_selector.value).isna()).any())
    return False

def update_summary():
    chunks = []
    for dataset, frame in RAW.items():
        nflag = sum(flagged(dataset,c) for c in frame.columns)
        chunks.append(f'<b>{dataset}</b>: {len(frame):,} rows, {len(frame.columns)} columns, '
                      f'{nflag} flagged columns')
    summary.value = ' &nbsp; | &nbsp; '.join(chunks)

def update_columns(*_):
    dataset = dataset_selector.value
    cols = [c for c in RAW[dataset].columns if search.value.lower() in str(c).lower()]
    if review_only.value:
        cols = [c for c in cols if flagged(dataset,c)]
    previous = column_selector.value
    column_selector.options = [(str(c),c) for c in cols]
    if previous in cols:
        column_selector.value = previous
    show_column()
    update_summary()

def show_column(*_):
    col = column_selector.value
    if col is None:
        column_info.value = '<i>No columns match. Uncheck “Review flagged only” to edit any column.</i>'
        type_selector.disabled = True
        return
    type_selector.disabled = False
    dataset = dataset_selector.value
    type_selector.unobserve(set_type, names='value')
    type_selector.value = settings[(dataset,col)]
    type_selector.observe(set_type, names='value')
    sample = normalize(RAW[dataset][col]).dropna().head(4).tolist()
    column_info.value = (f'<b>Suggested:</b> {suggestions[(dataset,col)]} &nbsp; '
                         f'<b>Current dtype:</b> {RAW[dataset][col].dtype} &nbsp; '
                         f'<b>Examples:</b> {str(sample)[:180]}')

def set_type(change):
    col = column_selector.value
    if col is not None:
        settings[(dataset_selector.value,col)] = change['new']
        # Don't rebuild the selector immediately; allow the user to finish editing.
        update_summary()

for control in (dataset_selector, search, review_only, format_selector):
    control.observe(update_columns, names='value')
column_selector.observe(show_column, names='value')

```

```python
def preview(_):
    with results:
        clear_output(wait=True)
        proposed, audit, issues = propose()
        print('Preview only — working DataFrames have NOT changed.')
        print(f'Columns to convert/retain: {len(audit):,}')
        if len(issues):
            print(f'\nREVIEW REQUIRED: {len(issues)} columns contain values that cannot be parsed.')
            display(issues)
            print('Change the column type, correct the source, or add genuine missing-value tokens.')
        else:
            print('No unparseable values detected for selected conversions.')
        display(audit.groupby(['Dataset','Action'],dropna=False).size().rename('Columns').to_frame())

def apply(_):
    global precip, stream, cleaning_report
    with results:
        clear_output(wait=True)
        proposed, audit, issues = propose()
        if len(issues):
            print('NOT APPLIED. Resolve these unparseable values first:')
            display(issues)
            return
        precip = proposed['Precipitation']
        stream = proposed['Stream']
        cleaning_report = audit
        print('Cleaning accepted. precip and stream updated; raw_precip/raw_stream preserved.')
        print('Audit available as cleaning_report.')
        display(audit.groupby(['Dataset','Action'],dropna=False).size().rename('Columns').to_frame())

preview_button.on_click(preview)
apply_button.on_click(apply)
update_columns()

dashboard = widgets.VBox([
    widgets.HTML('<h3>Data cleaning review</h3>'), summary,
    widgets.HBox([dataset_selector,review_only]),
    widgets.HBox([search,column_selector]),
    column_info, type_selector,
    widgets.HTML('<b>Global cleaning options</b>'),
    format_selector, extra_missing,
    widgets.HBox([preview_button,apply_button]), results
])
display(dashboard)
```

```python
import numpy as np
import ipywidgets as widgets
from IPython.display import display

# User enters the missing-value placeholder
missing_input = widgets.Text(
    value="",
    placeholder="Enter missing-value code",
    description="Missing value:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="350px")
)

replace_button = widgets.Button(
    description="Replace with NaN",
    button_style="primary"
)

output = widgets.Output()

def replace_missing(button):
    global precip, stream

    with output:
        output.clear_output()

        placeholder = missing_input.value.strip()

        # Handle numeric and text placeholders
        values = [placeholder]
        try:
            values.append(float(placeholder))
        except ValueError:
            pass

        # Count replacements before changing data
        precip_count = precip.isin(values).sum().sum()
        stream_count = stream.isin(values).sum().sum()

        # Replace placeholders with NaN
        precip = precip.replace(values, np.nan)
        stream = stream.replace(values, np.nan)

        print(f"Precipitation: {precip_count} values replaced")
        print(f"Stream: {stream_count} values replaced")

replace_button.on_click(replace_missing)

display(missing_input, replace_button, output)
```

```python
common_cols = precip.columns.intersection(stream.columns)

print("Shared columns:")
print(common_cols.tolist())
print("\nNumber of shared columns:", len(common_cols))

# I want the two files to match on year and month as the unique but shared values.
id_cols = ["Year", "Month"]

# Keep all columns except those I used to match. Those we don't need to double up on.
measurement_cols = [c for c in common_cols if c not in id_cols]

# Merge the files, but be sure to identify if the measure is from stream or precip.
paired = precip.merge(
    stream,
    on=id_cols,
    suffixes=("_precip", "_stream")
)

# This creates a dictionary to store keys (columns) and values (rows).
shared_counts = {}

# Then we count how many times they both have a non-*na* value.
for col in measurement_cols:
    shared_counts[col] = (
        paired[f"{col}_precip"].notna()
        & paired[f"{col}_stream"].notna()
    ).sum()

shared_counts

sum(shared_counts.values())

shared_counts_df = (
    pd.Series(shared_counts, name="paired_non_NA")
    .sort_values(ascending=False)
    .to_frame()
)

display(shared_counts_df)

print("Total paired non-NA cells:", shared_counts_df["paired_non_NA"].sum())
```

```python
# @title Use the dropdown menu to see a column's summary statistics.
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

legacy = pd.read_csv("/content/legacy.csv")

print("Dataset loaded:", legacy.shape)
print("Columns:", list(legacy.columns))
# Dropdown containing all columns in legacy
column_dropdown = widgets.Dropdown(
    options=list(legacy.columns),
    description="Select column:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="450px")
)

stats_output = widgets.Output()

def show_stats(change=None):
    with stats_output:
        stats_output.clear_output(wait=True)

        column = column_dropdown.value
        if column is None:
            return

        data = legacy[column]

        # Calculate descriptive statistics
        stats = data.describe()

        # Include missing values and total observations
        stats["Missing"] = data.isna().sum()
        stats["Total rows"] = len(data)
        stats["Missing %"] = round(data.isna().mean() * 100, 2)

        print(f"Descriptive statistics: {column}")
        display(stats.to_frame(name="Value"))

# Update automatically when a new column is selected
column_dropdown.observe(show_stats, names="value")

display(column_dropdown, stats_output)
show_stats()
```

```python
# @title Chemical Comparison { display-mode: "form" }

import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
import re

# Find chemical columns and their corresponding datasets
chemical_columns = {}

for col in legacy.columns:
    match = re.search(r"precipitation|precip|stream", str(col), re.I)

    if match:
        chemical = re.sub(
            r"precipitation|precip|stream", "", str(col),
            flags=re.I
        ).strip(" _-")

        source = "Precipitation" if "precip" in match.group().lower() else "Stream"

        chemical_columns.setdefault(chemical, {})[source] = col

# Only show chemicals found in BOTH datasets
chemicals = sorted(
    chem for chem, sources in chemical_columns.items()
    if "Precipitation" in sources and "Stream" in sources
)

chemical_dropdown = widgets.Dropdown(
    options=chemicals,
    description="Chemical:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="400px")
)

plot_output = widgets.Output()

def plot_chemical(change=None):
    with plot_output:
        plot_output.clear_output(wait=True)

        chemical = chemical_dropdown.value
        if chemical is None:
            print("No matching chemical columns found.")
            return

        columns = chemical_columns[chemical]

        precip_values = pd.to_numeric(
            legacy[columns["Precipitation"]], errors="coerce"
        )
        stream_values = pd.to_numeric(
            legacy[columns["Stream"]], errors="coerce"
        )

        # Use row number until a date column is identified
        x = range(len(legacy))

        plt.figure(figsize=(11, 5))

        plt.scatter(
            x, precip_values,
            label="Precipitation",
            color="royalblue",
            alpha=0.65,
            s=20
        )

        plt.scatter(
            x, stream_values,
            label="Stream",
            color="darkorange",
            alpha=0.65,
            s=20
        )

        plt.title(f"{chemical}: Precipitation vs Stream")
        plt.xlabel("Observation number")
        plt.ylabel("Chemical concentration")
        plt.legend()
        plt.grid(alpha=0.25)
        plt.tight_layout()
        plt.show()

chemical_dropdown.observe(plot_chemical, names="value")

display(chemical_dropdown, plot_output)
plot_chemical()
```

```python
# @title Chemical Correlation Heat Map { display-mode: "form" }

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import ipywidgets as widgets
from IPython.display import display

# Select which dataset to examine
source_dropdown = widgets.Dropdown(
    options=["Both", "Precipitation", "Stream"],
    value="Both",
    description="Data source:",
    style={"description_width": "initial"}
)

heatmap_output = widgets.Output()

def show_heatmap(change=None):
    with heatmap_output:
        heatmap_output.clear_output(wait=True)

        # Select columns based on source
        source = source_dropdown.value

        if source == "Both":
            cols = [
                c for c in legacy.columns
                if "precip" in str(c).lower()
                or "stream" in str(c).lower()
            ]
        else:
            term = "precip" if source == "Precipitation" else "stream"
            cols = [
                c for c in legacy.columns
                if term in str(c).lower()
            ]

        # Convert selected columns to numeric
        data = legacy[cols].apply(
            pd.to_numeric, errors="coerce"
        )

        # Exclude columns with insufficient observations
        data = data.loc[:, data.count() >= 3]

        if data.shape[1] < 2:
            print("Not enough numeric columns for a correlation heat map.")
            return

        # Pearson correlation
        corr = data.corr()

        # Plot
        plt.figure(figsize=(
            max(8, len(corr) * 0.7),
            max(6, len(corr) * 0.6)
        ))

        sns.heatmap(
            corr,
            annot=len(corr) <= 15,
            cmap="coolwarm",
            center=0,
            vmin=-1,
            vmax=1,
            square=True,
            linewidths=0.4,
            cbar_kws={"label": "Pearson correlation (r)"}
        )

        plt.title(f"Chemical Correlations — {source}")
        plt.tight_layout()
        plt.show()

source_dropdown.observe(show_heatmap, names="value")

display(source_dropdown, heatmap_output)
show_heatmap()
```
